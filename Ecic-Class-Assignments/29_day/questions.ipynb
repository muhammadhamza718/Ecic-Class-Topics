### Q1) Sum of positive number until negative

keep asking the user to enter numbers stop if the user enters a negative number print the sum of all positive numbers entered 

```python
total_sum = 0

while True:
    num = int(input("Enter your number: "))
    if num >= 0:
        total_sum += num
    else:
        break

print("Total Sum:", total_sum)
```

### Q2) print multiple of 3 and 5

print all number between 1 and 50 that are multiple of both 3 and 5 using a while loop 

```python
num = 1

while num <= 50:
    if num % 3 == 0 and num % 5 == 0:
        print(num)
    num += 1
```

### Q3) Count Digits in a Number

ask the user to enter a positive integer use a while loop and if statements to count how many digits in the number are even 

```python
num = int(input("Enter a positive integer: "))

count = 0
i = 1

if num > 0:
    while i <= num:
        if i % 2 == 0:
            print(i)
            count += 1
        i += 1

    print("Total even numbers:", count)
```

### Q4) Reverse a Number

Ask the user to input a number. Use a while loop and if statements to print the number in reverse. Only
reverse the digits if they are odd.

```python
num = int(input("Enter your number: "))
reverse = 0

while num > 0:
    digit = num % 10
    if digit % 2 != 0:
        reverse = reverse * 10 + digit
    num = num // 10

print(reverse)
```

Q5) Find Largest Number

Ask the user to enter numbers repeatedly. Stop when the user enters 0. Use a while loop and if
statements to find the largest number entered (ignore the 0).

```python
largest = 0

while True:
    num = int(input("Enter a number (0 to stop): "))
    if num == 0:
        break
    
    if num > largest:
        largest = num

print("The largest number is:", largest)
```
