## Library Book Checker

**Topics:** nested if/else, .strip(), .lower(), and/or, f-string

```python
book_name = input("Enter book name: ").strip().lower()
quantity = int(input("Enter number of books: "))
membership = input("Membership (student/teacher): ").strip().lower()

if quantity > 0 and quantity <= 10:
    if membership == "student":
        if quantity <= 3:
            if book_name == "research methods" or book_name == "advanced calculus" or book_name == "quantum physics":
                print(f"Sorry, '{book_name}' is for teachers only.")
            else:
                print(f"Approved! {quantity} book(s) of '{book_name}'")
        else:
            print(f"Students can only get 3 books")
    elif membership == "teacher":
        if quantity <= 10:
            print(f"Approved! {quantity} book(s) of '{book_name}'")
        else:
            print(f"Teachers can only get 10 books")
    else:
        print(f"Invalid membership. Enter 'student' or 'teacher'.")
else:
    print("Quantity must be between 1 and 10.")
```
