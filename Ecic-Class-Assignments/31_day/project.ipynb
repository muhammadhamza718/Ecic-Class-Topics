# PART 1: TREASURE ISLAND ADVENTURE GAME

### If/Else Decision Making + While Loop

A text-based adventure game that uses decision trees to create branching storylines.

### Flowchart

![Treasure Island flowchart](../../Ecic-&-Panaversity-Class-documents/static/treasure_island.png)

### Game Flow

```
Welcome to Treasure Island!
Your mission is to find the treasure.

Decision 1: left or right?
  → Left → continue
  → Right or anything else → Fall into a hole. Game Over.

Decision 2: swim or wait?
  → Wait → continue
  → Swim or anything else → Attacked by trout. Game Over.

Decision 3: Which door? (red/blue/yellow)
  → Red → Burned by fire. Game Over.
  → Blue → Eaten by beasts. Game Over.
  → Yellow → You Win!
  → Anything else → Game Over.

After each round → Play again? (yes/no)
  → yes → while loop goes back to start
  → no → break and end
```

### Requirements

- Use a **`while` loop** so the player can replay
- Use `if-elif-else` for decisions
- Use `.lower()` for case-insensitive input
- Ask **"Play again?"** at the end of each round

````python
# PART 1: Treasure Island Adventure Game

while True:
    print("=== WELCOME TO TREASURE ISLAND ===")
    print("Your mission is to find the treasure.")
    print()

    choice = input("Do you go left or right? ").lower()

    if choice == "left":
        print("You go left. The path opens to a river.")
        print()

        choice = input("Do you swim or wait? ").lower()

        if choice == "wait":
            print("You wait by the river. You hear mysterious sounds...")
            print()
            print("A cave reveals three doors: red, blue, yellow.")

            door = input("Which door do you choose? (red/blue/yellow) ").lower()

            if door == "yellow":
                print()
                print("⭐ CONGRATULATIONS! ⭐")
                print("You found the treasure!")
            elif door == "red":
                print("Burned by fire. Game Over.")
            elif door == "blue":
                print("Eaten by beasts. Game Over.")
            else:
                print("Invalid choice. Game Over.")
        else:
            print("Attacked by trout. Game Over.")
    else:
        print("You fall into a hole. Game Over.")

    print()
    again = input("Play again? (yes/no) ").lower()
    if again != "yes":
        print("Thanks for playing! Goodbye!")
        break
````

# PART 2: Rollercoaster Pricing System

## STAGE 0: HEIGHT CHECK ONLY

### Basic If/Else + While Loop

### Flowchart

![Rollercoaster Stage 0](../../Ecic-&-Panaversity-Class-documents/static/rollercoaster0.png)

```
Start
  ↓
While input is not a valid number → keep asking
  ↓
Height > 120cm?
  ├─ No → Can't ride
  └─ Yes → Can ride
```

### Requirements

- Use a **`while` loop** to keep asking for height until valid number
- Use `isdigit()` to check input before converting
- Use `break` to exit the loop on valid input
- If height > 120 → Can ride, else → Can't ride

````python
# STAGE 0: Height Check Only

print("=== ROLLERCOASTER BOOKING ===")

while True:
    h = input("What is your height (in cm)? ")
    if h.isdigit():
        h = int(h)
        break
    print("Invalid input! Please enter a number.")

if h > 120:
    print("✓ Great! You can ride this rollercoaster!")
else:
    print("✗ Sorry, you are too short to ride.")
    print("Minimum height required: 120cm")
````

## STAGE 1: HEIGHT + AGE-BASED PRICING

### Nested If/Else + Two While Loops

### Flowchart

```
Start
  ↓
While height input invalid → keep asking
  ↓
Height > 120cm?
  ├─ No → Can't ride
  └─ Yes → Can ride
            ↓
          While age input invalid → keep asking
            ↓
          Age?
          ├─ Under 12 → $5
          ├─ 12-18 → $7
          └─ Over 18 → $12
```

### Requirements

- **First `while` loop** for height validation
- **Second `while` loop** for age validation
- Use `if/elif/else` for age-based pricing

````python
# STAGE 1: Height + Age-Based Pricing

print("=== ROLLERCOASTER BOOKING ===")

while True:
    h = input("What is your height (in cm)? ")
    if h.isdigit():
        h = int(h)
        break
    print("Invalid input! Please enter a number.")

if h > 120:
    print("✓ Great! You can ride this rollercoaster!")
    print()

    while True:
        a = input("How old are you? ")
        if a.isdigit():
            a = int(a)
            break
        print("Invalid input! Please enter a number.")

    if a < 12:
        price = 5
        print("Age group: Child (Under 12)")
    elif a <= 18:
        price = 7
        print("Age group: Teenager (12-18)")
    else:
        price = 12
        print("Age group: Adult (18+)")

    print(f"Ticket price: ${price}")
else:
    print("✗ Sorry, you are too short to ride.")
    print("Minimum height required: 120cm")
````

## STAGE 2: HEIGHT + AGE + PHOTO ADD-ONS

### Complex Nested If/Else + Three While Loops

### Flowchart

![Rollercoaster Stage 2](../../Ecic-&-Panaversity-Class-documents/static/rollercoaster2.png)

```
Start
  ↓
While height input invalid → keep asking
  ↓
Height > 120cm?
  ├─ No → Can't ride
  └─ Yes → Can ride
            ↓
          While age input invalid → keep asking
            ↓
          Age?
          ├─ Less than 12 → +$5
          ├─ 12-18 → +$7
          └─ 18 or over → +$12
            ↓
          While photo input is not yes/no → keep asking
            ↓
          Want photos?
          ├─ No → total = base price
          └─ Yes → total = base price + $3
            ↓
          Display total bill
```

### Requirements

- Three `while` loops: height, age, photo choice
- Photo choice must be exactly `yes` or `no`
- Display itemized bill with ticket, photos, total

````python
# STAGE 2: Height + Age + Photo Add-ons

print("=== ROLLERCOASTER BOOKING ===")

while True:
    h = input("What is your height (in cm)? ")
    if h.isdigit():
        h = int(h)
        break
    print("Invalid input! Please enter a number.")

if h > 120:
    print("✓ Great! You can ride this rollercoaster!")
    print()

    while True:
        a = input("How old are you? ")
        if a.isdigit():
            a = int(a)
            break
        print("Invalid input! Please enter a number.")

    if a < 12:
        price = 5
        print("Age group: Child (Under 12)")
    elif a <= 18:
        price = 7
        print("Age group: Teenager (12-18)")
    else:
        price = 12
        print("Age group: Adult (18+)")

    print(f"Base ticket price: ${price}")
    print()

    while True:
        p = input("Do you want to buy photos? (yes/no) ").lower()
        if p in ["yes", "no"]:
            break
        print("Please type 'yes' or 'no'.")

    photo = 3 if p == "yes" else 0
    total = price + photo

    print()
    print("==== YOUR BILL ====")
    print(f"Ticket: ${price}")
    if photo > 0:
        print(f"Photos: +${photo}")
    else:
        print("Photos: $0")
    print("────────────")
    print(f"Total: ${total}")
else:
    print("✗ Sorry, you are too short to ride.")
    print("Minimum height required: 120cm")
````


### Requirements

- Three `while` loops: height, age, photo
- Senior ages 45-55 get **FREE ride ($0)** — check BEFORE `age > 18`
- Seniors still pay for photos (+$3)
- Show special celebration message for seniors

````python
# STAGE 3: Complete System with Senior Discount

print("=== ROLLERCOASTER BOOKING ===")

while True:
    h = input("What is your height (in cm)? ")
    if h.isdigit():
        h = int(h)
        break
    print("Invalid input! Please enter a number.")

if h > 120:
    print("✓ Great! You can ride this rollercoaster!")
    print()

    while True:
        a = input("How old are you? ")
        if a.isdigit():
            a = int(a)
            break
        print("Invalid input! Please enter a number.")

    if a < 12:
        price = 5
        print("Age group: Child (Under 12)")
    elif a <= 18:
        price = 7
        print("Age group: Teenager (12-18)")
    elif 45 <= a <= 55:
        price = 0
        print("Age group: Senior (45-55)")
        print("🎉 SPECIAL DISCOUNT! 🎉")
        print("Your ride is completely FREE!")
    else:
        price = 12
        print("Age group: Adult (18+)")

    print(f"Ticket price: ${price}")
    print()

    while True:
        p = input("Do you want to buy photos? (yes/no) ").lower()
        if p in ["yes", "no"]:
            break
        print("Please type 'yes' or 'no'.")

    photo = 3 if p == "yes" else 0
    total = price + photo

    print()
    print("==== YOUR BILL ====")
    if price == 0:
        print("Ticket: $0 (Senior Discount!)")
    else:
        print(f"Ticket: ${price}")
    if photo > 0:
        print(f"Photos: +${photo}")
    else:
        print("Photos: $0")
    print("────────────")
    print(f"Total: ${total}")
    if price == 0:
        print("Thank you for celebrating with us!")
else:
    print("✗ Sorry, you are too short to ride.")
    print("Minimum height required: 120cm")
````

## PROGRESSION CHALLENGE: COMBINING ALL STAGES

### Complete Interactive System with While + For Loops

- **Outer `while` loop** → process multiple customers
- **Inner `while` loops** → validate height, age, photo, and continue inputs
- **`for` loop** → print formatted summary report at the end

### Tracks:
- Total customers processed
- Total revenue collected
- Number of seniors who rode free
- Number of photo packages sold

````python
# PROGRESSION CHALLENGE: Complete Multi-Customer System

print("=== ROLLERCOASTER BOOKING SYSTEM ===")
print("Welcome to the ultimate theme park experience!")

count = 0
revenue = 0
seniors = 0
photos = 0

while True:
    count += 1
    print(f"\nCustomer {count}")
    print("──────────")

    # validate height
    while True:
        h = input("What is your height (in cm)? ")
        if h.isdigit():
            h = int(h)
            break
        print("Invalid input! Please enter a number.")

    if h <= 120:
        print("✗ Sorry, you are too short to ride.")
    else:
        print("✓ Great! You can ride this rollercoaster!")

        # validate age
        while True:
            a = input("How old are you? ")
            if a.isdigit():
                a = int(a)
                break
            print("Invalid input! Please enter a number.")

        if a < 12:
            price = 5
            print("Age group: Child (Under 12)")
        elif a <= 18:
            price = 7
            print("Age group: Teenager (12-18)")
        elif 45 <= a <= 55:
            price = 0
            seniors += 1
            print("Age group: Senior (45-55)")
            print("🎉 SPECIAL DISCOUNT! 🎉")
            print("Your ride is completely FREE!")
        else:
            price = 12
            print("Age group: Adult (18+)")

        print(f"Ticket price: ${price}")

        # validate photo choice
        while True:
            p = input("Do you want to buy photos? (yes/no) ").lower()
            if p in ["yes", "no"]:
                break
            print("Please type 'yes' or 'no'.")

        photo = 3 if p == "yes" else 0
        if photo > 0:
            photos += 1

        total = price + photo
        revenue += total

        print()
        print("==== YOUR BILL ====")
        if price == 0:
            print("Ticket: $0 (Senior Discount!)")
        else:
            print(f"Ticket: ${price}")
        if photo > 0:
            print(f"Photos: +${photo}")
        else:
            print("Photos: $0")
        print("────────────")
        print(f"Total: ${total}")
        if price == 0:
            print("Thank you for celebrating with us!")

    # validate continue choice
    while True:
        again = input("\nProcess another customer? (yes/no) ").lower()
        if again in ["yes", "no"]:
            break
        print("Please type 'yes' or 'no'.")

    if again == "no":
        break

# for loop to print summary
print("\n=== SESSION SUMMARY ===")
data = [
    ("Total Customers Processed", count),
    ("Total Revenue", f"${revenue}"),
    ("Photo Packages Sold", photos),
    ("Senior Free Rides", seniors),
]
for label, value in data:
    print(f"{label}: {value}")

if count > 0:
    avg = revenue / count
    print(f"Average Revenue per Customer: ${avg:.2f}")

print("Thank you for visiting!")
````
