# Kashmir Solidarity Event Manager (Simple Version)

This is a simple program to manage a Kashmir Day event.

### What it does:
1. Asks for the camp name and budget.
2. Manages a list of cities.
3. Calculates the total cost of items.
4. Checks if you have enough money.
5. Displays the final details.

```python
print("==============================")
print("KASHMIR DAY EVENT MANAGER")
print("==============================")

camp_name = input("Enter Camp Name: ")
budget = float(input("Enter Budget: "))

cities = ["Muzaffarabad", "Mirpur", "Rawalakot"]
print("Cities:", cities)

new_city = input("Enter a new city: ")
cities.append(new_city)
print("Total Cities:", len(cities))

items = ["Banners", "Flags", "Sound System"]
prices = [1500, 800, 5000]

total_cost = sum(prices)
print("Total Cost:", total_cost)

if budget >= total_cost:
    print("Status: You have enough money")
    if budget - total_cost > 5000:
        print("You can buy a Digital Screen now")
        items.append("Digital Screen")
        prices.append(4500)
else:
    print("Status: Not enough money")
    items.pop()
    prices.pop()
    print("One item removed to save money")

print("Final Camp Details")
print("Name:", camp_name)
print("Total Items:", len(items))
print("Total Expense:", sum(prices))
print("Money Left:", budget - sum(prices))
```
