## STAGE 3: COMPLETE SYSTEM WITH SENIOR DISCOUNT

### Advanced Nested If/Else + Three While Loops

### Flowchart

![Rollercoaster Stage 3](../../Ecic-&-Panaversity-Class-documents/static/rollercoaster3.png)

```
Start
  ↓
While height input invalid → keep asking
  ↓
Height > 120cm?
  ├─ No → Can't ride
  └─ Yes → Can ride
            ↓
          While age input invalid → keep asking
            ↓
          Age?
          ├─ Less than 12 → +$5
          ├─ 12-18 → +$7
          ├─ 45-55 (Senior) → +$0 ⭐
          └─ 18 or over → +$12
            ↓
          While photo input is not yes/no → keep asking
            ↓
          Want photos?
          ├─ No → total = base price
          └─ Yes → total = base price + $3
            ↓
          Display total bill
```

### Requirements

- Three `while` loops: height, age, photo
- Senior ages 45-55 get **FREE ride ($0)** — check BEFORE `age > 18`
- Seniors still pay for photos (+$3)
- Show special celebration message for seniors

````python
print("=== ROLLERCOASTER BOOKING ===")

while True:
    h = input("What is your height (in cm)? ")
    if h.isdigit():
        h = int(h)
        break
    print("Invalid input! Please enter a number.")

if h > 120:
    print("✓ Great! You can ride this rollercoaster!")
    print()

    while True:
        a = input("How old are you? ")
        if a.isdigit():
            a = int(a)
            break
        print("Invalid input! Please enter a number.")

    if a < 12:
        price = 5
        print("Age group: Child (Under 12)")
    elif a <= 18:
        price = 7
        print("Age group: Teenager (12-18)")
    elif 45 <= a <= 55:
        price = 0
        print("Age group: Senior (45-55)")
        print("🎉 SPECIAL DISCOUNT! 🎉")
        print("Your ride is completely FREE!")
    else:
        price = 12
        print("Age group: Adult (18+)")

    print(f"Ticket price: ${price}")
    print()

    while True:
        p = input("Do you want to buy photos? (yes/no) ").lower()
        if p in ["yes", "no"]:
            break
        print("Please type 'yes' or 'no'.")

    photo = 3 if p == "yes" else 0
    total = price + photo

    print()
    print("==== YOUR BILL ====")
    if price == 0:
        print("Ticket: $0 (Senior Discount!)")
    else:
        print(f"Ticket: ${price}")
    if photo > 0:
        print(f"Photos: +${photo}")
    else:
        print("Photos: $0")
    print("────────────")
    print(f"Total: ${total}")
    if price == 0:
        print("Thank you for celebrating with us!")
else:
    print("✗ Sorry, you are too short to ride.")
    print("Minimum height required: 120cm")
````
