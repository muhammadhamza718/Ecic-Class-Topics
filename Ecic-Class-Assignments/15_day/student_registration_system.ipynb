# 📚 Student Course Registration System

## Scenario Description

You are building a **Student Course Registration System** for a university. The system helps students register for courses based on their eligibility, preferences, and availability.

---

## 🎯 System Requirements

### 1. **Student Information** (Variables & Data Types)
- Store student's **name** (string)
- Store student's **roll number** (integer)
- Store student's **CGPA** (float)
- Store student's **year of study** (integer: 1, 2, 3, or 4)
- Store whether student **has paid fees** (boolean)
- Store whether student **has cleared previous semester** (boolean)

### 2. **String Operations**
- Extract student's **first name** and **last name** from full name using slicing
- Display student **initials** using indexing
- Create a **student ID** by concatenating roll number and first 3 letters of name
- Use **negative indexing** to get the last character of the name

### 3. **Available Courses** (Lists, Tuples, Sets)
- Create a **list** of available courses: `["Python Programming", "Data Science", "Web Development", "Machine Learning", "AI Ethics"]`
- Create a **tuple** for course details (course_name, credits, max_students)
  - Example: `("Python Programming", 3, 50)`
- Create a **set** of prerequisite courses a student has already completed
  - Example: `{"Introduction to Programming", "Mathematics"}`

### 4. **Eligibility Checks** (Operators & Boolean)
Use **comparison operators** to check:
- Is CGPA >= 2.5?
- Is fees_paid == True?
- Is cleared_previous_semester == True?
- Is year_of_study <= 4?

Use **logical operators** (and, or, not) to combine conditions

### 5. **Registration Logic** (If-Elif-Else & Nested Conditions)

#### **Level 1: Basic Eligibility**
- **IF** student has NOT paid fees → Display "❌ Cannot register. Please pay fees first."
- **ELIF** student has NOT cleared previous semester → Display "❌ Cannot register. Clear previous semester first."
- **ELSE** → Proceed to Level 2

#### **Level 2: Academic Standing** (Nested If-Else)
If basic eligibility passed:
- **IF** CGPA >= 3.5 → Student is "Excellent" → Can register for ANY course including advanced ones
  - **Nested IF**: Check if "Machine Learning" is in the course list
    - If yes → Allow registration
    - Else → Suggest "Data Science" as alternative
- **ELIF** CGPA >= 2.5 and CGPA < 3.5 → Student is "Good" → Can register for standard courses only
  - **Nested IF**: Check if course is in basic courses list (first 3 courses)
- **ELSE** CGPA < 2.5 → Student is "Need Improvement" → Can only register for 1 course

#### **Level 3: Year-Based Restrictions** (More Nested Conditions)
- **IF** year_of_study == 1 → Can only take courses from index 0-1 of the list
- **ELIF** year_of_study == 2 → Can take courses from index 0-2
- **ELIF** year_of_study == 3 → Can take courses from index 0-4
- **ELSE** (year 4) → Can take all courses
