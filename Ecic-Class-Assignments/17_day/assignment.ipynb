```python
number1 = int(input("Enter your first number: "))
number2 = int(input("Enter your second number: "))
number3 = int(input("Enter your third number: "))
number4 = int(input("Enter your fourth number: "))
number5 = int(input("Enter your fifth number: "))

full_list = [number1, number2, number3, number4, number5]
sum_of_First_two = full_list[0] + full_list[1]
sum_of_last_two = full_list[-1] + full_list[-2]

first_half = full_list[0:2]
second_half = full_list[-2:]

if full_list[0] > full_list[-1]:
    if sum_of_First_two > sum_of_last_two:
        print(f"First half {first_half}")
    else:
        print(f"Second half {second_half}")
else:
    if sum_of_First_two < sum_of_last_two:
        print(f"First half {first_half}")
    else:
        print(f"Second half {second_half}")
```
