## Python Sets

Symmetric diff without using built-in methods

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

sym_diff = (set1 - set2) | (set2 - set1)
print("Symmetric Difference:", sym_diff)
```

## Sets with nested if, ifelse

check if an element exists

```python
my_set = {10, 20, 30, 40, 50}
num = 25

if num in my_set:
    print(f"{num} is present in the set.")
elif num > max(my_set):
    print(f"{num} is greater than all elements in the set.")
else:
    print(f"{num} does not exist in the set and is less than the maximum element.")
```

### Nested if:
