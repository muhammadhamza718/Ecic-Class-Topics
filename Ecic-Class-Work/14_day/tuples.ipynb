## Python Tuples

Tuples:

- Tuples are used to store multiple items in a single variable
- Tuples are ordered
- Tuples are immutable (cannot change or update)
- used with round brackets

Example:

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple)
```

Uses:

- Use with dictionaries as a key

```python
locations = {(40.7128, -74.0060): "New York", (34.0522, -118.2437): "Los Angeles"}
print(locations)
```

Length:

```python
thistuple = ("apple", "banana", "cherry")
print(len(thistuple))
```

One item tuple:

```python
thistuple = ("apple",)
print(type(thistuple))
```

Using Tuple():

```python
thistuple = tuple(("apple", "banana", "cherry"))
print(thistuple)
```

```python
tup = tuple("Python")
print(tup)
```
