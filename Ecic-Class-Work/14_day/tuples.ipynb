## Python Tuples

Tuples:

- Tuples are used to store multiple items in a single variable
- Tuples are ordered
- Tuples are immutable (cannot change or update)
- used with round brackets

Example:

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple)
```

Uses:

- Use with dictionaries as a key

```python
locations = {(40.7128, -74.0060): "New York", (34.0522, -118.2437): "Los Angeles"}
print(locations)
```

Length:

```python
thistuple = ("apple", "banana", "cherry")
print(len(thistuple))
```

One item tuple:

```python
thistuple = tuple("apple")
print(thistuple)
print(type(thistuple))
```

Using Tuple():

```python
thistuple = tuple(("apple", "banana", "cherry"))
print(thistuple)
```

```python
tup = tuple("Python")
print(tup)
```

Concatenation of tuple:

- we can add only datatype
- list & tuple cannot be combined  

```python
tuple1 = (0, 1, 2, 3)
tuple2 = ("python", "learning")
Tuple3 = tuple1 + tuple2

print(Tuple3)
```

Deleting a tuple:

- generate an error 
- tuple are immutable
- we cannot delete individual items

```python
tup = (1, 2, 3, 4)
del tup
print(tup)
```

```python
my_list = list(thistuple)
thistuple = tuple(my_list)
print(thistuple)
```

-> How to modify a tuple

- create a list, modify & convert back to tuple


```python
my_tuple = ("apple", "banana", "cherry")
my_list = list(my_tuple)
my_list.append("orange")
my_tuple = tuple(my_list)
print(my_tuple)
```
