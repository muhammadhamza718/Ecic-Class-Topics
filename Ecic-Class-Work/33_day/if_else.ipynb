### **📌 Topic 1:**  `if / else`

**Core mindset:** An if/else is a decision gate. Your program hits a fork in the road and must choose a path based on a condition that is either `True` or `False`.

**Example:**

```
if condition:
    # path A
else:
    # path B
```

### **✅ 10 Basic Questions**

**Q1.** Write a program that takes a number and prints `"Positive"` if it's greater than 0, otherwise prints `"Non-positive"`.

````python
number = int(input("Enter your number: "))

if number > 0:
    print("Positive")
else:
    print("Non-Positive")
````

**Q2.** A store gives a 10% discount if the total bill is above ₹500. Write a program that takes the bill amount and prints the final amount to pay.

````python
bill = int(input("Enter your number: "))

if bill > 500:
    discount = bill * 0.10
    total_amount = bill - discount
else:
    total_amount = bill

print(f"Final amount to pay: ₹{total_amount:.2f}")
````

**Q3.** Write a program that checks if a person is eligible to vote. (Age must be 18 or above.)

````python
age = 18

if age >= 18:
    print("Eligible for voting!")
else:
    print("Not eligible for voting!")
````

**Q4.** You have a variable `is_raining = True`. Print `"Take an umbrella"` if it's raining, otherwise print `"Enjoy the weather"`.

````python
is_raining = True 

if is_raining:
    print("Take an umbrella")
else:
    print("Enjoy the weather")
````

**Q5.** Write a program that takes two numbers and prints which one is larger. If they're equal, print `"Both are equal"`.

````python
number_1 = int(input("Enter your first number: "))
number_2 = int(input("Enter your second number: "))

if number_1 > number_2:
    print(f"first number is greater than second number: {number_1} > {number_2}")
elif number_2 > number_1:
    print(f"second number is greater than first number: {number_2} > {number_1}")
else: 
    print(f"Both first/second number are equal: {number_1} = {number_2}")
````

**Q6.** A user enters a password. If it matches `"secret123"`, print `"Access granted"`, otherwise print `"Access denied"`.

````python
password = input("Enter your password: ").strip().lower()

if password == "secret123":
    print("Access granted!")
else:
    print("Access denied!")
````
