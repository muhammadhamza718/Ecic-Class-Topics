### **📌 Topic 1:**  `if / else`

**Core mindset:** An if/else is a decision gate. Your program hits a fork in the road and must choose a path based on a condition that is either `True` or `False`.

**Example:**

```
if condition:
    # path A
else:
    # path B
```

### **✅ 10 Basic Questions**

**Q1.** Write a program that takes a number and prints `"Positive"` if it's greater than 0, otherwise prints `"Non-positive"`.

````python
number = int(input("Enter your number: "))

if number > 0:
    print("Positive")
else:
    print("Non-Positive")
````

**Q2.** A store gives a 10% discount if the total bill is above ₹500. Write a program that takes the bill amount and prints the final amount to pay.

````python
bill = float(input("Enter your number: "))

if bill > 500:
    discount = bill * 0.10
    total_amount = discount - bill
else:
    total_amount = bill

print(f"Final amount to pay: ₹{total_amount:.2f}")
````
