# **For Loops**

## 01) Example:

```python
fruits = ["apple", "banana", "cherry"]

for i in range(5):
    print(fruits)
    print(*fruits)
    print(*fruits, sep=", ")
```

## 02) Example:

- Using range() to access indices

```python
fruits = ["apple", "banana", "cherry"]

for i in range(len(fruits)):
    print(f"index {i} contains {fruits[i]}")
```

## 03) Example:

- Creating a list of sequence

```python
squares = [2**2 for i in range(10)]
print(squares)
```

## 04) Example:

```python
for i in range(1, 6):
    plus_ten = i + 10
    minus_five = i - 5
    print(f"Number: {i}, plus ten: {plus_ten}, minus five: {minus_five}")
```
