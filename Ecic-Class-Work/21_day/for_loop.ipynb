# **For Loops**

## 01) Example:

```python
fruits = ["apple", "banana", "cherry"]

for i in range(5):
    print(fruits)
    print(*fruits)
    print(*fruits, sep=", ")
```

## 02) Example:

- Using range() to access indices

```python
fruits = ["apple", "banana", "cherry"]

for i in range(len(fruits)):
    print(f"index {i} contains {fruits[i]}")
```
