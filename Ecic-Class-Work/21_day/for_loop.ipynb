# **For Loops**

## 01) Example:

```python
fruits = ["apple", "banana", "cherry"]

for i in range(5):
    print(fruits)
    print(*fruits)
    print(*fruits, sep=", ")
```

## 02) Example:

- Using range() to access indices

```python
fruits = ["apple", "banana", "cherry"]

for i in range(len(fruits)):
    print(f"index {i} contains {fruits[i]}")
```

## 03) Example:

- Creating a list of sequence

```python
squares = [2**2 for i in range(10)]
print(squares)
```

## 04) Example:

```python
for i in range(1, 6):
    plus_ten = i + 10
    minus_five = i - 5
    print(f"Number: {i}, plus ten: {plus_ten}, minus five: {minus_five}")
```

```python
total = 0

for i in range(1, 11):
    total += i
    print(f"the summ of numbers from 1 to 10 is: {total}")
```

```python
result = 0

for i in range(1, 6):
    if i%2 ==0:
        result = result - i
        print(f"Subtracted {i}: Current total is {result}")
    else:
        result = result + i
        print(f"Added {i}: Current total is {result} ")
```

```python
for i in range(1, 4):
    print(f"{i} sequenced is {i**2}")
```

```python
for i in range(1, 6):
    start_num = i
    i = i + 10
    after_plus = i
    i = i -5
    after_minus = i
    print(f"Start: {start_num} | Added: {after_plus} | Minus five: {after_minus}")
```
