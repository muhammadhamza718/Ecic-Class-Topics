## **Program 1: Conditional String Manipulation**

### Description
This program takes a string as input and performs different operations based on the length and
properties of the string.

Input: 
- A single string from the user.

Processing Logic:
- If string length \> 5:  
	- Condition 1:  
		- If first half of the string equals the reverse of the second half  
			- Output: "Palindrome Half"  
	- Condition 2:  
		- Else if the string is in uppercase  
			- Output: string with all vowels removed  
	- Else:  
		- Swap the first half and second half of the string  
- Else (length ≤ 5):  
	- Print the string twice, excluding the first character


```python
text = input("Enter some text: ")

length = len(text)

if length > 5:
    mid = length // 2
    first_half = text[:mid]
    second_half = text[mid:]
    
    if first_half == second_half[::-1]:
        print("Palindrome half detected!")
    elif text.upper():
        result = text.replace("A", "").replace("E", "").replace("I", "").replace("O", "").replace("U", "")
        print("Text without vowels:", result)
    else:
        part1 = text[:mid]
        part2 = text[mid:]
        print(part1 + part2)
else:
    short_version = text[1]
    print("Short version:", short_version)
```
