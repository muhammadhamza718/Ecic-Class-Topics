### **📌 Topic 2: Nested if / else**

**Core mindset:** Sometimes one decision isn't enough — the answer to the first question opens up another question. That's nesting. Think of it like a flowchart inside a flowchart.

```python
if condition_1:
    if condition_2:
        # path A
    else:
        # path B
else:
    # path C
```

**✅ 10 Basic Questions**

**Q1.** Check if a number is positive. If it is, further check if it's even or odd. If it's not positive, just print `"Negative or Zero"`.

```python
number = 0

if number > 0:
    if number % 2 == 0:
        print("Even")
    else:
        print("Odd")
else:
    print("Negative or Zero")
```
