### Key points:

- Tuples are used to store multiple items in a single variabke
- Tuples are ordered and immutable 
- used with () round brackets
- allow duplicate values 
- tuples are indexed

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple)
```

```python
thistuple = ("apple")
print(thistuple)
```

```python
thistuple = ("apple", 34, True)
print(thistuple)
```

```python
thistuple = tuple(("apple", "banana", "cherry"))
print(thistuple)
```

```python
thistuple = ("apple", "banana", "cherry")
del thistuple
print(thistuple)
```

### Slicing

```python
thistuple = ("apple", "mango", "cherry", "banana", "melon", "orange")
print(thistuple[-2])
```

```python
thistuple = ("apple", "mango", "cherry", "banana", "melon", "orange")
print(thistuple[2:5])
```

### Tuple with if statement

```python
thistuple = ("apple", "banana", "cherry")

if "apple" in thistuple:
    print("apple is in the variable")
else:
    print("apple is not in the variable")
```
