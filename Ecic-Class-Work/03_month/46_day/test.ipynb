```python
def two_number(x, y):
    return x - y

two_number(3, -3)
```

```python
def three_number(x, y, z):
    number = 0
    if number < x:
        number = x
    if number < y:
        number = y
    if number < z:
        number = z
    return number

result = three_number(2, 4, 12)
print(result)
```

```python
def drop_float_digit(number):
    if number % 2 == 0:
        return float(number)
    else:
        if number == float(number):
            x1 = int(number)
            result_number = number - x1
            return result_number
result = drop_float_digit(3.7)
print(result)
```

```python
def mix_word(x, y):
    return x + y

mix_word("Ham", "za")
```

```python
def minimal_number(number_list=[10, 10, 6, 10, 10]):
    number = number_list[0]
    for i in number_list:
        if i < number:
            number = i
    return number

minimal_number()
```

```python
def average_func(x, y):
    average = (x + y) / 2
    return average

average_func(5, 10)
```
