```python
def multiply(a, b):
    return a * b

print(multiply(4, 5))  # 20 aana chahiye
```

```python
def calculate_bill(price, discount=10):
    result = price - price * discount / 100
    return result

# calculate_bill(1000)

result = calculate_bill(1000)
print(result)
```

```python
def average(*args):
    args = list(args)
    if not args:
        print("No values provided")
        return 0.0
    return sum(args) / len(args)

# average(10, 20, 30)
# average(5, 15)
# average(1, 2, 3, 4, 5)
average()
```

```python
def build_profile(**kwargs):
    for k, v in kwargs.items():
        print(f"{k}: {v}")

build_profile(name="Hamza", course="Python", level="Beginner")
```
