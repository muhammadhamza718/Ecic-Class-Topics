```python
def dict_count():
    sentence = input("Enter any sentence: ")
    words = sentence.split()
    word_counts = {}
    position_number = 1
    
    for word in words:
        word_counts[word] = position_number
        position_number += 1
            
    return word_counts

counts = dict_count()
print(counts)
```

```python
def count_same_character():
    name = "Muhammad hamza"
    
    char_counts = {}
    
    for char in name:
        if char in char_counts:
            char_counts[char] += 1
        else:
            char_counts[char] = 1
            
    return char_counts

counts = count_same_character()
print(counts)
```
