## Multi-Level Number Analysis

### Q1: Multi-Level Number Analysis

**Question:**
Write a program that takes a number and:

* **If positive:**
  * Check if it is even or odd.
  * Then check if it is greater than 50 or not.
  * Output combined result.
* **If negative:**
  * Check if divisible by both 3 and 5.
  * Otherwise check if divisible by only one of them.
* **If zero:**
  * Print a special message.

```python
num = 45

if num > 0:
    if num % 2 == 0:
        if num > 50:
            print("Positive Even and Greater than 50")
        else:
            print("Positive Even and Less or Equal to 50")
    else:
        if num > 50:
            print("Positive Odd and Greater than 50")
        else:
            print("Positive Odd and Less or Equal to 50")

elif num < 0:
    if num % 3 == 0 and num % 5 == 0:
        print("Negative and divisible by both 3 and 5")
    elif num % 3 == 0 or num % 5 == 0:
        print("Negative and divisible by either 3 or 5")
    else:
        print("Negative and not divisible by 3 or 5")

else:
    print("Number is Zero")
```

### Q2: Smart Grade Evaluation

**Question:**
Create a grading system:

* Marks ≥ 90 → A+
* 80–89 → A
* 70–79 → B
* 60–69 → C
* **Below 60:**
  * If marks ≥ 50 → "Pass but needs improvement"
  * Else → "Fail"

```python
marks = 58

if marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
elif marks >= 60:
    print("C")
else:
    if marks >= 50:
        print("Pass but needs improvement")
    else:
        print("Fail")
```

### Q3: Advanced Largest Finder

**Question:**
Find the largest among three numbers. If two or more numbers are equal and largest, print a special message.

```python
a = 20
b = 20
c = 15

if a >= b and a >= c:
    if a == b or a == c:
        print("A is largest but tied")
    else:
        print("A is largest")
elif b >= a and b >= c:
    if b == a or b == c:
        print("B is largest but tied")
    else:
        print("B is largest")
else:
    if c == a or c == b:
        print("C is largest but tied")
    else:
        print("C is largest")
```

### Q4: Year Validation System

**Question:**
Check if a year is valid and leap:

* Year must be ≥ 1000
* If not valid → print error
* If valid → check leap year rules

```python
year = 2024

if year >= 1000:
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print("Valid Leap Year")
    else:
        print("Valid but Not Leap Year")
else:
    print("Invalid Year")
```

### Q5: Enhanced Calculator with Validation

**Question:**
Build a calculator:

* Perform +, -, *, /
* **If division:** Handle division by zero
* **If invalid operator:** Show error

```python
a = 10
b = 0
op = "/"

if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    if b != 0:
        print(a / b)
    else:
        print("Error: Division by zero")
else:
    print("Invalid operator")
```

## DICTIONARY + IF-ELSE (Advanced Level)

### Q1: Student Performance Analyzer

**Question:**
Using a dictionary:

* If marks ≥ 75 → "Excellent"
* 50–74 → "Average"
* **Below 50:**
  * If attendance ≥ 80 → "Conditionally Pass"
  * Else → "Fail"

```python
student = {"name": "Ali", "marks": 48, "attendance": 85}

if student["marks"] >= 75:
    print("Excellent")
elif student["marks"] >= 50:
    print("Average")
else:
    if student["attendance"] >= 80:
        print("Conditionally Pass")
    else:
        print("Fail")
```

### Q2: Product Pricing Logic

**Question:**
Using dictionary:

* If price > 5000: Apply 10% discount
* Else if price > 2000: Apply 5% discount
* Else: No discount

```python
product = {"name": "Phone", "price": 6000}

if product["price"] > 5000:
    discount = product["price"] * 0.10
elif product["price"] > 2000:
    discount = product["price"] * 0.05
else:
    discount = 0

final_price = product["price"] - discount
print(final_price)
```

### Q3: Secure Login System

**Question:**
Check login:

* Correct username and password → success
* Wrong password → specific message
* Wrong username → different message

```python
user = {"username": "admin", "password": "1234"}

input_user = "admin"
input_pass = "1111"

if input_user == user["username"]:
    if input_pass == user["password"]:
        print("Login successful")
    else:
        print("Wrong password")
else:
    print("Username not found")
```

### Q4: Employee Salary System

**Question:**
Using dictionary:

* If salary ≥ 70000: Bonus = 20%
* 50000–69999: Bonus = 10%
* **Below 50000:**
  * If experience ≥ 5 → Bonus 5%
  * Else → No bonus

```python
employee = {"name": "Ahmed", "salary": 48000, "experience": 6}

if employee["salary"] >= 70000:
    bonus = employee["salary"] * 0.20
elif employee["salary"] >= 50000:
    bonus = employee["salary"] * 0.10
else:
    if employee["experience"] >= 5:
        bonus = employee["salary"] * 0.05
    else:
        bonus = 0

print("Bonus:", bonus)
```

### Q5: Weather Alert System

**Question:**
Using dictionary:

* **If temp > 35:**
  * If humidity > 70 → "Heat Alert"
  * Else → "Hot"
* If temp between 20–35 → "Normal"
* Below 20 → "Cold"

```python
weather = {"city": "Karachi", "temp": 36, "humidity": 75}

if weather["temp"] > 35:
    if weather["humidity"] > 70:
        print("Heat Alert")
    else:
        print("Hot")
elif weather["temp"] >= 20:
    print("Normal")
else:
    print("Cold")
```
