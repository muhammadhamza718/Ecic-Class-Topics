- Dictionary are used to store data values in key value pair.
- A dictionary is a collection which is ordered, changable & dont allow duplicates
- Dictionary are written with only brackets & have key & value 

```python
thisdict = {"brand": "BMW"}

print(thisdict)
```

```python
thisdict = {
    "name": "Hamza", 
    "age": 18, 
    "age": 18, 
    "education": "Intermediate", 
    "car": ["BMW", "Toyota", "Mercedes"],
    "gender": "Male"
}
print(len(thisdict))
print(thisdict["age"])
```

```python
thisdict = {"name": "Hamza", "age": 18, "field": "software engineer"} 

print(thisdict)
```

```python
student = {"name": "hamza", "age": 18}

for key, value in student.items():
    print(key, value)
```
