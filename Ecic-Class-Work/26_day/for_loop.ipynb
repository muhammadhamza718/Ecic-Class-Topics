```python
raw_prices = ["$10.50", "$20.00", "$5.25"]
clean_prices = []

for price in raw_prices: 
    numeric_price = float(price.strip("$"))
    clean_prices.append(numeric_price)
print(clean_prices)
```

```python
transaction = [120, 45, 900, 10, 150]
premium_sales = []

for amount in transaction:
    if amount > 100:
        premium_sales.append(amount)

print(premium_sales)
```

```python
montly_revenue = [4000, 5200, 3800, 6100]
total_yearly = 0
for revenue in montly_revenue:
    total_yearly += revenue
print(f"Total Revenue: {total_yearly}")
```

```python
original_prices = [100, 50, 20, 150]
discount = 0.90

for price in original_prices:
    sale_price = price * discount
    print(f"Discount: ${price} -> Sale Price: ${sale_price}")
```
