### 📌 Topic 3: Slicing

**Core mindset:** Slicing is how you zoom into a piece of data. You're not changing the original — you're just looking through a window into it.

`data[start : stop : step]`

- `start` → where to begin (inclusive)
- `stop` → where to end (exclusive)
- `step` → how many to skip

✅ 10 Basic Questions

**Q1.** Given `name = "Alexander"`, extract just `"Alex"` using slicing.
