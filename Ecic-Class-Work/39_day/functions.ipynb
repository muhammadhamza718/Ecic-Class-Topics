# Functions ⚙:

A func is a block of code that perform a specific task.

### Syntax:

```
def function_name(): 
    #code
```

### Example 1:

````python
def greet():
    print("Hello Python!")

greet()
````

### Example 2:

````python
def greet(name):
    print("Hello", name)

greet("Hamza")
````

````python
def percentage(total_marks, total_gained_marks):
    return total_marks / total_gained_marks * 100

result = percentage(50, 100)
int(result)

print(f"{result}%")
````

````python
def even_odd(num):
    if num % 2 == 0:
        print("even")
    else:
        print("odd")

even_odd(6)
````

````python
def greater_less(a, b):
    if a > b:
        print(f"{a} is greater than {b}")
    else:
        print(f"{b} is greater than {a}")

greater_less(6, 4)
````

````python
def celsius(temp):
    return (temp - 32) * 5 / 9

celsius(77)
````

````python
def fahrenheit_to_celsius(fahrenheit):
  return (fahrenheit - 32) * 5 / 9

print(fahrenheit_to_celsius(77))
print(fahrenheit_to_celsius(95))
print(fahrenheit_to_celsius(50))
````

````python
def grade(percentage):
    if percentage >= 80:
        return "A+"
    elif percentage >= 70:
        return "A"
    elif percentage >= 60:
        return "B"
    elif percentage >= 50:
        return "C"
    elif percentage >= 40:
        return "D"
    elif percentage >= 30:
        return "E"
    else:
        return "F"


def percentage(total_marks, gained_marks):
    return (gained_marks / total_marks) * 100


result = percentage(100, 50)
result = round(result, 2)

final_grade = grade(result)

print(f"Your total percentage is {result}%")
print(f"Your grade is {final_grade}")
````
