# While Loop

- A while loop is used to repeat a block of code as long as condition is True
- If check the condition before each iteration 
- If condition become false, The loop stops

### Syntax of while loop

```
while condition:
    # Block of code 
while - keyword
- condition - boolean Exp (True / False)
- Indentation is compulsary in python
```

How while loop works.

- check the condition 
- if True - execute the code inside the loop 
- Go back & check the condition argue 
- Repeat until condition become false

````python
i = 1
while i <= 5:
    print(i)
    i += 1
````

````python
num = int(input("Enter your name: "))

while num > 0:
    print(f"Positive number: {num}")
    num -= 1

while num < 0:
    print(f"Negative number: {num}")
    num += 1
````

### Using break in while loop:

````python
i = 1
while i <= 10:
    if i == 5:
        break
    print(i)
    i += 1
````

### Using break in while loop

````python
i = 1
while i <= 10:
    if i == 5:
        continue
    print(i)
    i += 1
````

````python
balance = 100000

while True:
    print("\n Select your option:")
    print("\n 1. Deposit")
    print("\n 2. Withdraw")
    print("\n 3. Check Balance")
    print("\n 4. Exit")
    print()
    print()
    print()

    choice = int(input("Choose any option (1-4):"))

    if choice == 1:
        deposit_amount = float(input("Enter your deposit amount: "))
        balance += deposit_amount
        print("Deposit the amount successfully!")
    elif choice == 2:
        withdraw_amount = float(input("Enter your withdraw amount: "))
        balance -= withdraw_amount
        print("Withdraw the amount successfully!")
    elif choice == 3:
        print(f"Your bank Balance is {balance}")
    elif choice == 4:
        break
    else:
        print("Invalid Option")
````
