```python
number = int(input("Enter your number: "))

if number > 0:
    if number % 2 == 0:
        print("Even")
    else:
        print("Odd")
elif number < 0:
    print(abs(number))
else: 
    print("Number is zero")
```

```python

s = input("Enter any string: ")

length = len(s)

vowels = set('aeiou')
count = 0
for char in s.lower():
    if char in vowels:
        count += 1

total_vowels = count
print(f"Total vowels in the string: {total_vowels}")
print(f"Reverse string {s[::-1]}")


middle_half = length // 2
first_half = s[:middle_half]
second_half = s[middle_half::]

if first_half == second_half:
    print("Palindrome")
else:
    print("Not palindrome")

print(s[::2])
```
