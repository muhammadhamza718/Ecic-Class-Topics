# **Topic: For Loop**

### Intro:

A for loop in python is used to repeat a block of code multiple times. It is mainly used when we know how many times we want to run the loop or when we want to iterate over a sequence such as (string, list, tuple) or range of numbers.

### Uses:

- To repeat tasks automatically 
- To work with collection 
- To perform calculations on multiple values
- To make program efficient & organized 

### Example:

### 1. A for loop with a string 


```python
word = "Python"
for letter in word:
    print(let ter)
```


### 2. loop with a list

```python
colors = ["red", "yellow", "pink"]
for color in colors:
    print(color)
```

### 3. Loop with a tuple

```python
fruits = ("apple", "banana", "cherry")
for fruit in fruits:
    print(fruit)
```

### Using Range() function:

- The range function generates sequence of numbers

### Example:

```python
for i in range(5):
    print(i)
```

### For loop with if statement:

```python
for i in range(1, 11):
    # print(i)
    if i % 3 == 0:
        print(i)
```

### Using break in for loop

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```
